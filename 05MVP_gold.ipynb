# Camada Gold

```sql
-- Define o catálogo e o schema da camada Gold,
-- onde serão criadas as tabelas utilizadas nas análises.
USE CATALOG mvp;
USE SCHEMA gold;
```

```sql
-- Cria (ou substitui) o Data Mart que apresenta
-- a diversidade de produtos energéticos produzidos por UF.
CREATE OR REPLACE TABLE diversidade_uf AS

SELECT
    uf,

    -- Conta a quantidade de produtos energéticos distintos
    -- produzidos por cada Unidade da Federação.
    COUNT(DISTINCT produto) AS qtd_produtos

FROM silver.producao_energetica

-- Considera somente registros em que houve produção.
WHERE producao > 0

-- Agrupa os resultados por Unidade da Federação.
GROUP BY uf
```

```sql
-- Consulta os dados do Data Mart para validar
-- o resultado da agregação por Unidade da Federação.
SELECT *
FROM diversidade_uf
```


### Catálogo de Dados - **_gold.diversidade_uf_**


| Coluna           | Tipo   | Descrição                                                                                                                                                                             |
| ---------------- | ------ | ------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **uf**           | string | Unidade da Federação considerada na análise da diversidade da produção energética.                                                                                                    |
| **qtd_produtos** | int    | Quantidade de produtos energéticos distintos produzidos pela Unidade da Federação durante o período analisado (2014–2023), considerando apenas registros com produção maior que zero. |


```sql
-- Cria (ou substitui) o Data Mart que apresenta
-- a diversidade de produtos energéticos por UF em cada ano.
CREATE OR REPLACE TABLE diversidade_ano AS

SELECT
    ano,
    uf,

    -- Conta a quantidade de produtos energéticos distintos
    -- produzidos por cada UF em cada ano.
    COUNT(DISTINCT produto) AS qtd_produtos

FROM silver.producao_energetica

-- Considera somente registros em que houve produção.
WHERE producao > 0

-- Agrupa os resultados por ano e Unidade da Federação.

GROUP BY
    ano,
    uf
```

```sql
-- Consulta os dados do Data Mart para validar
-- a diversidade calculada por ano e Unidade da Federação.
SELECT *
FROM gold.diversidade_ano
```

### Catálogo de dados - **_gold.diversidade_ano_**

 | Coluna           | Tipo   | Descrição                                                                                                                                                       |
| ---------------- | ------ | --------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **ano**          | int    | Ano de referência da produção energética, compreendido entre 2014 e 2023.                                                                                       |
| **uf**           | string | Unidade da Federação considerada na análise da diversificação anual da produção energética.                                                                     |
| **qtd_produtos** | int    | Quantidade de produtos energéticos distintos produzidos pela Unidade da Federação no respectivo ano, considerando apenas registros com produção maior que zero. |

```sql
-- Cria (ou substitui) o Data Mart utilizado para analisar
-- a expansão territorial dos produtos energéticos ao longo dos anos.
CREATE OR REPLACE TABLE expansao_produtos AS

SELECT

    ano,

    produto,

    -- Conta a quantidade de Unidades da Federação
    -- que registraram produção de cada produto em cada ano.
    COUNT(DISTINCT uf) AS qtd_ufs

FROM silver.producao_energetica

-- Considera somente registros em que houve produção.
WHERE producao > 0

-- Agrupa os resultados por ano e produto energético.
GROUP BY

    ano,

    produto
```

```sql
-- Consulta os dados do Data Mart para validar
-- a quantidade de UFs produtoras por produto e ano.
SELECT *
FROM gold.expansao_produtos
```

### Catálogo de Dados - **_expansao_produtos_**

| Coluna      | Tipo   | Descrição                                                                                                                                                       |
| ----------- | ------ | --------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **ano**     | int    | Ano de referência da produção energética, compreendido entre 2014 e 2023.                                                                                       |
| **produto** | string | Produto energético analisado conforme classificação da ANP.                                                                                                     |
| **qtd_ufs** | int    | Quantidade de Unidades da Federação que registraram produção do respectivo produto no ano analisado, considerando apenas registros com produção maior que zero. |
