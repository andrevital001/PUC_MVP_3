# Carregamento dos arquivos extraídos do Github no Volume dados_anp

```sql
-- Define o catálogo e o schema onde será criado o Volume
-- utilizado para armazenar os arquivos da camada Staging.
USE CATALOG mvp;
USE SCHEMA staging;
```

```sql
-- Cria um Volume para armazenar os arquivos CSV
-- obtidos da base pública da ANP.
CREATE VOLUME dados_anp
```

```sql
%python

import os
import requests

# Define o caminho do Volume onde os arquivos serão armazenados.
volume_path = "/Volumes/mvp/staging/dados_anp"

# Cria o diretório caso ele ainda não exista.
os.makedirs(volume_path, exist_ok=True)

# Lista dos arquivos CSV que serão obtidos do repositório público.
arquivos = [
    "Tabela2.10.csv",
    "Tabela2.13.csv",
    "Tabela2.9.csv",
    "Tabela4.1.csv",
    "Tabela4.10.csv",
    "Tabela4.17.csv"
]

# URL base do repositório GitHub contendo os arquivos.
base_url = "https://raw.githubusercontent.com/andrevital001/PUC/main/"


# Percorre todos os arquivos da lista para realizar o download.
    
    
    
    

for arquivo in arquivos:
    # Monta a URL completa do arquivo.
    url = base_url + arquivo
    # Define o caminho de destino dentro do Volume.
    destino = os.path.join(volume_path, arquivo)
    
    print(f"Baixando {arquivo}...")

    # Realiza o download do arquivo.
    response = requests.get(url)

    # Interrompe a execução caso ocorra algum erro durante o download.
    response.raise_for_status()
   
    # Salva o arquivo no Volume em formato binário.
    with open(destino, "wb") as f:
        f.write(response.content)

print("\nTodos os arquivos foram carregados para o Volume.")
```
