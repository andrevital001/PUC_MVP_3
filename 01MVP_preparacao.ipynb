# Criação de Catálogo **_mvp_** e Schemas a serem utilizados

```sql
%sql
-- Remove o catálogo caso ele já exista.
-- O parâmetro CASCADE exclui também todos os schemas e objetos contidos nele.
DROP CATALOG IF EXISTS mvp CASCADE
```

```sql
%sql
-- Cria o catálogo que armazenará todos os objetos utilizados no projeto.
CREATE CATALOG mvp
```

```sql

-- Define o catálogo "mvp" como contexto padrão
-- para as próximas operações.
USE CATALOG mvp
```

```sql
-- Remove o schema padrão criado automaticamente
-- junto com o catálogo, evitando sua utilização no projeto.
DROP SCHEMA IF EXISTS mvp CASCADE
```

### Criação dos schemas:
* staging
* bronze
* silver
* gold
* analise


```sql
-- Cria o schema Staging, responsável por armazenar
-- os arquivos obtidos da fonte de dados antes do processamento.
CREATE SCHEMA staging
```

```sql
-- Remove o schema Bronze caso já exista,
-- permitindo recriar toda a estrutura do projeto.
DROP SCHEMA IF EXISTS bronze CASCADE
```

```sql
-- Cria o schema Bronze para armazenar
-- os dados brutos consolidados.
CREATE SCHEMA bronze
```

```sql
-- Remove o schema Silver caso já exista.
DROP SCHEMA IF EXISTS silver CASCADE
```

```sql
-- Cria o schema Silver, responsável pelo tratamento,
-- limpeza e padronização dos dados.
CREATE SCHEMA silver
```

```sql
-- Remove o schema Gold caso já exista.
DROP SCHEMA IF EXISTS gold CASCADE
```

```sql
-- Cria o schema Gold, onde serão armazenadas
-- as tabelas analíticas utilizadas nas análises.
CREATE SCHEMA gold
```

```sql
-- Remove o schema utilizado para armazenar
-- os resultados das consultas analíticas.
DROP SCHEMA IF EXISTS analise CASCADE
```

```sql
-- Cria o schema Analise, destinado ao armazenamento
-- das consultas e resultados gerados durante a etapa de análise.
CREATE SCHEMA analise
```
