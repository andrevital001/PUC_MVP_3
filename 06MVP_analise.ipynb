**Introdução da Análise**

**Pergunta 1**

_Quais Unidades da Federação apresentam maior diversidade de produtos energéticos produzidos ao longo da série histórica?_


```sql
-- Consulta a quantidade de produtos energéticos distintos
-- identificados em cada Unidade da Federação.
SELECT
    uf,
    qtd_produtos
FROM mvp.gold.diversidade_uf

-- Ordena as UFs da maior para a menor diversidade de produtos.
ORDER BY qtd_produtos DESC,
         uf;
```

```sql
-- Define o catálogo e o schema utilizados para armazenar
-- os resultados das consultas da etapa de análise.
USE CATALOG mvp;
USE SCHEMA analise;
```

```sql
-- Cria (ou substitui) a tabela que armazena
-- o resultado da análise de diversidade por UF.
CREATE OR REPLACE TABLE analise1 AS

SELECT
    uf,
    qtd_produtos
FROM mvp.gold.diversidade_uf

-- Organiza os resultados da maior para a menor
-- quantidade de produtos energéticos distintos.
ORDER BY qtd_produtos DESC,
         uf;
```

**Pergunta 2**

_Existe tendência de diversificação da matriz produtiva estadual entre 2014 e 2023?_

```sql
-- Calcula a média da quantidade de produtos energéticos
-- produzidos pelas UFs em cada ano da série histórica.
SELECT
    ano,
    AVG(qtd_produtos) media_produtos
FROM mvp.gold.diversidade_ano

-- Agrupa os registros por ano para permitir
-- o acompanhamento da evolução da média ao longo do período.
GROUP BY ano

-- Ordena os resultados em ordem cronológica.
ORDER BY ano;
```

```sql
-- Cria (ou substitui) a tabela que armazena
-- o resultado da análise da diversidade média anual.
CREATE OR REPLACE TABLE analise2 AS

SELECT
    ano,

    -- Calcula a média da quantidade de produtos
    -- registrados pelas UFs em cada ano.
    AVG(qtd_produtos) media_produtos
FROM mvp.gold.diversidade_ano

-- Agrupa os registros por ano.
GROUP BY ano

-- Organiza os resultados em ordem cronológica.
ORDER BY ano;
```

**Pergunta 3**


_Quais produtos apresentaram maior expansão territorial ao longo dos anos?_

```sql
-- Consulta a quantidade de UFs produtoras
-- de cada produto energético em cada ano.
SELECT
    ano,
    produto,
    qtd_ufs
FROM mvp.gold.expansao_produtos

-- Organiza os registros por produto e,
-- dentro de cada produto, do ano mais recente para o mais antigo.
ORDER BY produto, ano DESC;
```

```sql
-- Compara a quantidade de UFs produtoras de cada produto
-- no início e no final da série histórica.
SELECT
    produto,

    -- Obtém a quantidade de UFs produtoras em 2014.
    -- Caso não exista registro para o produto no ano, considera zero.
    COALESCE(
        MAX(CASE WHEN ano = 2014 THEN qtd_ufs END),
        0
    ) AS ufs_2014,

    -- Obtém a quantidade de UFs produtoras em 2023.
    -- Caso não exista registro para o produto no ano, considera zero.
    COALESCE(
        MAX(CASE WHEN ano = 2023 THEN qtd_ufs END),
        0
    ) AS ufs_2023,

    -- Calcula a variação absoluta da quantidade de UFs produtoras
    -- entre 2014 e 2023.
    COALESCE(
        MAX(CASE WHEN ano = 2023 THEN qtd_ufs END),
        0
    )
    -
    COALESCE(
        MAX(CASE WHEN ano = 2014 THEN qtd_ufs END),
        0
    ) AS crescimento_absoluto

FROM mvp.gold.expansao_produtos

-- Agrupa os registros por produto energético.
GROUP BY produto


-- Ordena os produtos da maior expansão territorial
-- para a maior redução no período.
ORDER BY crescimento_absoluto DESC, produto;
```

```sql

-- Cria (ou substitui) a tabela que armazena
-- a evolução anual da presença territorial de cada produto.
CREATE OR REPLACE TABLE analise3 AS

SELECT
    ano,
    produto,
    qtd_ufs
FROM mvp.gold.expansao_produtos

-- Organiza os registros por produto e ano.
ORDER BY produto, ano DESC;
```

```sql
-- Cria (ou substitui) a tabela que armazena
-- a variação territorial de cada produto entre 2014 e 2023.
CREATE OR REPLACE TABLE analise3_1 AS

SELECT
    produto,

    -- Quantidade de UFs produtoras no início da série histórica.
    COALESCE(
        MAX(CASE WHEN ano = 2014 THEN qtd_ufs END),
        0
    ) AS ufs_2014,


    -- Quantidade de UFs produtoras no final da série histórica.
    COALESCE(
        MAX(CASE WHEN ano = 2023 THEN qtd_ufs END),
        0
    ) AS ufs_2023,

    -- Calcula a expansão ou redução territorial
    -- ocorrida entre 2014 e 2023.
    COALESCE(
        MAX(CASE WHEN ano = 2023 THEN qtd_ufs END),
        0
    )
    -
    COALESCE(
        MAX(CASE WHEN ano = 2014 THEN qtd_ufs END),
        0
    ) AS crescimento_absoluto

FROM mvp.gold.expansao_produtos

-- Agrupa os resultados por produto energético.
GROUP BY produto

-- Ordena os produtos de acordo com a variação territorial observada.
ORDER BY crescimento_absoluto DESC, produto;
```
